# Value Set Optimization

Compact enumerated code lists into include/exclude rules, then inspect the result before publishing.

```python
import os
import medterm4ds as mt

DB = os.environ.get("MEDTERM4DS_DB", "/mnt/d/medterm4ds/data/umls_local.duckdb")
terms = mt.connect(DB, memory_profile="low")
```

```python
codes = ["E11.40", "E11.41", "E11.42", "E11.43", "E11.44"]
result = terms.optimize("ICD10CM", codes, include_codes=True)
result.to_dict(include_codes=True)
```

```python
rules = [rule.to_dict(include_codes=True) for rule in result.rules]
rules
```

```python
assert result.original_count == len(codes)
assert result.optimized_count <= result.original_count
assert rules[0]["include_source"] == "ICD10CM"
terms.close()
```
