# Patient-Friendly Names and Mapping Review

Use DataFrames to keep review fields visible while generating patient-friendly labels and source-to-target mappings.

```python
import os
import medterm4ds as mt

DB = os.environ.get("MEDTERM4DS_DB", "/mnt/d/medterm4ds/data/umls_local.duckdb")
terms = mt.connect(DB, memory_profile="low")
```

```python
codes = ["E11.9", "E11.40", "E11.42"]
friendly = terms.patient_friendly_df("ICD10CM", codes)
friendly[["source", "code", "name", "friendly_source", "match_type", "match_depth", "matched_via"]]
```

```python
mapping = terms.map_df(
    "ICD10CM",
    ["E11.9"],
    target_sources=["SNOMEDCT_US"],
    max_depth=2,
)
mapping[["source", "code", "target_source", "target_code", "target_display", "relationship", "match_type", "match_depth", "matched_via"]]
```

```python
assert "matched_via" in friendly.columns
assert "match_type" in mapping.columns
assert len(mapping) >= 1
terms.close()
```
