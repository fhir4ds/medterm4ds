# Terminology Lookup, Search, and Inventory

This notebook shows the notebook-first medterm4ds workflow: connect once, inspect sources, search names, and look up exact codes.

```python
import os
import medterm4ds as mt

DB = os.environ.get("MEDTERM4DS_DB", "/mnt/d/medterm4ds/data/umls_local.duckdb")
terms = mt.connect(DB, memory_profile="low")
```

```python
stats = terms.source_stats_df(["ICD10CM", "SNOMEDCT_US", "RXNORM", "LNC", "CVX"])
stats
```

```python
search = terms.search_df("diabetes", sources=["ICD10CM", "SNOMEDCT_US"], limit=10)
search[["source", "code", "name", "tty", "match_type"]]
```

```python
lookup = terms.lookup_df([mt.CodeRef("ICD10CM", "E11.9"), mt.CodeRef("LNC", "4548-4")])
lookup[["source", "code", "name", "cui", "tty"]]
```

```python
assert not stats.empty
assert "E11.9" in set(lookup["code"])
assert not search.empty
terms.close()
```
