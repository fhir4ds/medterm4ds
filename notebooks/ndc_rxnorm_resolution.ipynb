# NDC to RxNorm and Historical Code Resolution

Resolve package-level NDCs and obsolete codes while preserving original identifiers for downstream review.

```python
import os
import medterm4ds as mt

DB = os.environ.get("MEDTERM4DS_DB", "/mnt/d/medterm4ds/data/umls_current.duckdb")
terms = mt.connect(DB, memory_profile="low")
```

```python
ndc = terms.resolve("NDC", "0002-0821-01")
ndc.to_dict()
```

```python
historical = terms.resolve("ICD10CM", "OLD")
historical.to_dict()
```

```python
df = terms.resolve_df([mt.CodeRef("NDC", "0002-0821-01"), mt.CodeRef("ICD10CM", "OLD")])
df[["source", "code", "resolved_source", "resolved_code", "status", "match_type", "matched_via"]]
```

```python
assert ndc.resolved == mt.CodeRef("RXNORM", "12345")
assert historical.status == "replaced"
assert set(df["status"]) == {"ndc_resolved", "replaced"}
terms.close()
```
